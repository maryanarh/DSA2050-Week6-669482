```python
#Part 1 – Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
```

```python
#Part 2 – Load the Dataset
df = pd.read_csv("DSA2050_Week6_Sales_Data.csv")
df.head()

print("Shape:", df.shape)
print()
df.info()
```

```python
#Part 3 – Basic Data Audit
print("Missing values:")
display(df.isna().sum())

print("Duplicate rows:", df.duplicated().sum())

display(df.describe(include="all"))

df["Sales_KES"].describe()

plt.figure(figsize=(8,4))
plt.hist(df["Sales_KES"], bins=25)
plt.title("Distribution of Sales")
plt.xlabel("Sales (KES)")
plt.ylabel("Frequency")
plt.show()
```

# Interpretation
- Sales range from ≈17,805 KES to ≈295,825 KES, with a mean of ≈169,252 KES and a roughly symmetric distribution centered around 168,000 KES. No obvious outliers or data-entry anomalies are present, so no extreme values were removed

#Part 4 – Define the Analytical Task
**Business question:** Can available business characteristics be used to estimate sales?

**Analytical task:** supervised regression.

**Target:** Sales_KES.

**Features:** `AdvertisingSpend_KES`, `StoreVisits`, `DiscountPercent`, `Region`, `CustomerSegment`, `ProductCategory`

```python
#Part 5 – Separate Features and Target
X = df.drop(columns="Sales_KES")
y = df["Sales_KES"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Part 6 – Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))
```

```python
#Part 7 – Identify Feature Types
numeric_features = [
    "AdvertisingSpend_KES",
    "StoreVisits",
    "DiscountPercent"
]

categorical_features = [
    "Region",
    "CustomerSegment",
    "ProductCategory"
]
```

```python
#Part 8 – Leakage-Safe Preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
#Part 9 – Build Baselines
mean_baseline = DummyRegressor(strategy="mean")
median_baseline = DummyRegressor(strategy="median")

mean_baseline.fit(X_train, y_train)
median_baseline.fit(X_train, y_train)

mean_pred = mean_baseline.predict(X_test)
median_pred = median_baseline.predict(X_test)
```

#Part 10 – Evaluation Metrics
- MAE: average absolute prediction error; lower is better.

- RMSE: penalizes large errors more heavily than MAE; lower is better.

- R²: how much variation in the target is explained by the model; higher is generally better.

```python
def evaluate_regression(y_true, y_pred, name):
    return pd.DataFrame({
        "Model": [name],
        "MAE": [mean_absolute_error(y_true, y_pred)],
        "RMSE": [np.sqrt(mean_squared_error(y_true, y_pred))],
        "R2": [r2_score(y_true, y_pred)]
    })

mean_results = evaluate_regression(y_test, mean_pred, "Mean Baseline")
median_results = evaluate_regression(y_test, median_pred, "Median Baseline")

pd.concat([mean_results, median_results], ignore_index=True)
```

```python
# Part 11 – Linear Regression
linear_model = Pipeline([
    ("preprocess", preprocessor),
    ("model", LinearRegression())
])

linear_model.fit(X_train, y_train)
linear_pred = linear_model.predict(X_test)

linear_results = evaluate_regression(
    y_test, linear_pred, "Linear Regression"
)
linear_results
```

# Interpretation
Linear Regression MAE ≈13,758 vs baseline ≈39,197 — lower.

RMSE ≈17,470 vs ≈48,716 — lower.

R² ≈0.870 vs ≈0 — higher.

The model clearly adds useful predictive value

```python
#Part 12 – Ridge Regression
ridge_model = Pipeline([
    ("preprocess", preprocessor),
    ("model", Ridge(alpha=1.0))
])

ridge_model.fit(X_train, y_train)
ridge_pred = ridge_model.predict(X_test)

ridge_results = evaluate_regression(
    y_test, ridge_pred, "Ridge Regression"
)
ridge_results
```

```python
#Part 13 – Shallow Decision Tree
tree_model = Pipeline([
    ("preprocess", preprocessor),
    ("model", DecisionTreeRegressor(
        max_depth=4,
        min_samples_leaf=8,
        random_state=RANDOM_STATE
    ))
])

tree_model.fit(X_train, y_train)
tree_pred = tree_model.predict(X_test)

tree_results = evaluate_regression(
    y_test, tree_pred, "Decision Tree"
)
tree_results
```

```python
#Part 14 – Compare All Models
comparison = pd.concat([
    mean_results,
    median_results,
    linear_results,
    ridge_results,
    tree_results
], ignore_index=True)

comparison.sort_values("MAE")
```

##Explanation
Ridge Regression achieved the lowest MAE (≈13,714 KES), lowest RMSE (≈17,459 KES), and highest R² (≈0.8705), with Linear Regression almost identical (MAE ≈13,758, R² ≈0.8703).

The Decision Tree was weaker (MAE ≈23,776, R² ≈0.648), while the Mean and Median baselines were worst (MAE ≈39,000, R² ≈0).

All fitted models clearly beat the baselines.

```python
#Part 15 – Cross-Validation on Training Data Only
scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

models = {
    "Linear Regression": linear_model,
    "Ridge Regression": ridge_model,
    "Decision Tree": tree_model
}

rows = []
for name, model in models.items():
    scores = cross_validate(
        model, X_train, y_train,
        cv=5,
        scoring=scoring
    )
    rows.append({
        "Model": name,
        "CV_MAE": -scores["test_MAE"].mean(),
        "CV_RMSE": -scores["test_RMSE"].mean(),
        "CV_R2": scores["test_R2"].mean()
    })

cv_results = pd.DataFrame(rows)
cv_results
```

#Questions

###**Which model performs best on average**
- Ridge performs best on average (CV_MAE ≈15,971, CV_R² ≈0.848), with Linear Regression virtually tied (15,977 / 0.848). The Decision Tree is worse (24,498 / 0.651).

###**Are the CV results consistent with the test results?**
 - The CV ranking matches the test-set ranking (Ridge ≈ Linear > Tree > Baselines), so the results are consistent.

###**Why is training-data cross-validation safer than repeatedly checking the final test set?**
 - Because every time you look at the test set and adjust your model, you leak information from the test set into model selection. Keeping the test set untouched preserves its value as an unbiased final estimate.

```python
#Part 16 – Residual Analysis
#Residual = Actual − Predicted.
#Positive residuals mean the model under-predicted;
#negative residuals mean the model over-predicted.

residuals = y_test - linear_pred

residual_df = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": linear_pred,
    "Residual": residuals.values
})

residual_df.head()

plt.figure(figsize=(8,5))
plt.scatter(linear_pred, residuals)
plt.axhline(0)
plt.title("Linear Regression Residual Plot")
plt.xlabel("Predicted Sales (KES)")
plt.ylabel("Residual: Actual - Predicted")
plt.show()
```

# Interpretation:
- Residuals are centred around zero, which is good. There is a slight funnel shape (wider spread at higher predicted values), suggesting mild heteroscedasticity, but no strong pattern. A handful of large residuals exist > 40,000 KES, meaning the model occasionally under- or over-predicts by a wide margin.

```python
# Part 17 – Actual vs Predicted
plt.figure(figsize=(8,5))
plt.scatter(y_test, linear_pred)
plt.title("Actual vs Predicted Sales")
plt.xlabel("Actual Sales (KES)")
plt.ylabel("Predicted Sales (KES)")
plt.show()
```

#Interpretation:
- If predictions were perfect, points would lie close to an imaginary diagonal line where Actual = Predicted. Here the points cluster roughly along the diagonal, confirming a reasonable fit, but with visible scatter at both ends.

```python
#Part 18 – Inspect Largest Errors
error_analysis = X_test.copy()
error_analysis["ActualSales"] = y_test
error_analysis["PredictedSales"] = linear_pred
error_analysis["Residual"] = (
    error_analysis["ActualSales"] - error_analysis["PredictedSales"]
)
error_analysis["AbsoluteError"] = error_analysis["Residual"].abs()

error_analysis.sort_values(
    "AbsoluteError", ascending=False
).head(10)
```

##Questions:

### **Do the largest errors share characteristics?**
 - They tend to have very low or very high advertising spend and store visits, i.e., edge cases.

### **Are they concentrated in one region, segment, or category?**
 - They are spread across regions and segments, but many involve SME customers and Electronics, where sales tend to be very high or very low.

### **What additional data might explain them?**
 - Competitor activity, seasonality, promotions, footfall quality, and customer income levels could explain the residual variance.

```python
#Part 19 – Error by Customer Segment
segment_errors = error_analysis.groupby(
    "CustomerSegment"
).agg(
    Observations=("AbsoluteError", "size"),
    MAE=("AbsoluteError", "mean")
).reset_index()

segment_errors
```

##Explanation:
 - Retail has the highest MAE (≈15,561 KES), followed by Corporate (≈14,300), while SME has the lowest (≈10,481). Sample sizes are similar (18–27 per segment), so the differences are not just noise.


```python
#Part 20 – Check for Overfitting
tree_train_pred = tree_model.predict(X_train)

print(
    "Shallow Tree Training MAE:",
    round(mean_absolute_error(y_train, tree_train_pred), 2)
)
print(
    "Shallow Tree Test MAE:",
    round(mean_absolute_error(y_test, tree_pred), 2)
)

```

```python
#Part 21 – Demonstrate deep tree train vs test MAE
deep_tree = Pipeline([
    ("preprocess", preprocessor),
    ("model", DecisionTreeRegressor(
        random_state=RANDOM_STATE
    ))
])

deep_tree.fit(X_train, y_train)

deep_train_pred = deep_tree.predict(X_train)
deep_test_pred = deep_tree.predict(X_test)

print(
    "Deep Tree Training MAE:",
    round(mean_absolute_error(y_train, deep_train_pred), 2)
)
print(
    "Deep Tree Test MAE:",
    round(mean_absolute_error(y_test, deep_test_pred), 2)
)
```

```python
#Part 22 - depth tuning
depths = [2, 4, 6, None]
depth_results = []

for d in depths:
    m = Pipeline([
        ("preprocess", preprocessor),
        ("model", DecisionTreeRegressor(
            max_depth=d,
            min_samples_leaf=8,
            random_state=RANDOM_STATE
        ))
    ])
    m.fit(X_train, y_train)
    pred = m.predict(X_test)
    depth_results.append({
        "max_depth": d,
        "Test_MAE": mean_absolute_error(y_test, pred),
        "Test_RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "Test_R2": r2_score(y_test, pred)
    })

pd.DataFrame(depth_results)
```

#Explanation:
- An almost-perfect training MAE (near 0) with a much worse test MAE is the classic signature of overfitting. The unconstrained tree memorises the training data, including its noise, and therefore fails to generalise to unseen data. A low training error alone does not indicate a better model — it indicates a model that has stopped learning general patterns and started memorising specifics.

#Part 23 – Final Recommendation
- The strongest baseline was the Mean Baseline with a test MAE of ≈39,197 KES. Ridge Regression was the best candidate (MAE ≈13,714, RMSE ≈17,459, R² ≈0.8705), essentially tied with Linear Regression.

- It beat the baseline by ≈25,500 KES MAE — a very large improvement. In business terms, predictions are on average off by about 13,700 KES per record.

- Residuals are centred on zero with a slight funnel shape at higher predicted values.

- Performance differed by segment: Retail had the highest MAE, SME the lowest.

- The shallow tree showed no strong overfitting (train MAE 18,030 vs test 23,776); the deep tree was clearly overfit (train 0, test 26,798).

- I would not deploy immediately — I would first add features (seasonality, competitor activity, promotions) and validate on a larger, more recent sample.

#Part 24 – Limitations
- Small dataset (320 rows): limits model stability and generalisation; a few hundred more rows could change the coefficient estimates meaningfully.

- Limited predictors: the model ignores seasonality, competitor pricing, promotions, macroeconomic conditions, and store-level attributes that strongly influence sales.

- Prediction is not causation: the model shows association, not cause-and-effect. Increasing advertising spend may not directly raise sales because both may be driven by a third factor.

- Teaching/synthetic data: the dataset may not reflect real-world noise and distributional quirks.

- Data drift: relationships learned from past data may not hold in future periods.
